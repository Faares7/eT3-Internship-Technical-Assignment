```python
import pandas as pd

df = pd.read_excel('/content/Combined_Test_Data.xlsx')

ID = df['ID'].tolist()
Area = df['Area'].tolist()
Priority = df['Priority'].tolist()
Weight = df['Package Weight (kg)'].tolist()
```

```python

deliveredPackages = []
totalTrips = 0
totalWeightDelivered = 0

while len(ID) > 0:

    if Weight[0] > 10.0:
        print(f"Item {ID[0]} is over 10kg and cannot be delivered")
        ID.pop(0)
        Area.pop(0)
        Priority.pop(0)
        Weight.pop(0)
        continue

    currentPackages = []
    deliveryWeight = 0
    tripArea = ""
    added = True

    while added:
        added = False

        best = -1
        minPriority = 999
        minWeight = 999

        for i in range(len(ID)):

            if deliveryWeight + Weight[i] <= 10.0:

                if Priority[i] < minPriority:
                    minPriority = Priority[i]
                    minWeight = Weight[i]
                    best = i

                elif Priority[i] == minPriority:

                    if tripArea != "" and Area[i] == tripArea and Area[best] != tripArea:
                        minWeight = Weight[i]
                        best = i

                    elif tripArea != "" and Area[best] == tripArea and Area[i] != tripArea:
                        pass

                    else:
                        if Weight[i] < minWeight:
                            minWeight = Weight[i]
                            best = i

        if best != -1:
            if tripArea == "":
                tripArea = Area[best]

            deliveryWeight += Weight[best]
            currentPackages.append(ID[best])

            ID.pop(best)
            Area.pop(best)
            Priority.pop(best)
            Weight.pop(best)

            added = True

    deliveredPackages.append(currentPackages)
    totalTrips += 1
    totalWeightDelivered += deliveryWeight
    print(f"Delivered Packages {currentPackages} to {tripArea} - Total Weight: {deliveryWeight}kg")

print('All Items Have Been Delivered')
print("-" * 30)

# Extension: Business Summary
print("TRIP EFFICIENCY SUMMARY")
print(f"Total Trips Required: {totalTrips}")
if totalTrips > 0:
    maxPossibleCapacity = totalTrips * 10.0
    utilization = (totalWeightDelivered / maxPossibleCapacity) * 100
    print(f"Overall Fleet Utilization: {utilization:.1f}%")
```
